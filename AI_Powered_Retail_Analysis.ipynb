```python
df_original = df.copy()

df_clean = df.drop_duplicates().copy()

print("Original rows:", len(df_original))
print("Rows after removing duplicates:", len(df_clean))
print("Duplicate rows removed:", len(df_original) - len(df_clean))
```

```python
df_clean["Description"] = (
    df_clean["Description"].fillna("Unknown Product")
)

print("Missing Description:",
      df_clean["Description"].isna().sum())

print("Missing Customer ID:",
      df_clean["Customer ID"].isna().sum())

print("Remaining rows:", len(df_clean))
```

```python
df_sales = df_clean[
    (df_clean["Quantity"] > 0) &
    (df_clean["Price"] > 0) &
    (df_clean["InvoiceDate"].notna())
].copy()

# Calculate total sales for each transaction line
df_sales["Sales"] = (
    df_sales["Quantity"] * df_sales["Price"]
)

print("Total cleaned rows:", len(df_clean))
print("Valid sales rows:", len(df_sales))
print("Rows excluded from sales analysis:",
      len(df_clean) - len(df_sales))

print("\nTotal Sales Value:", round(df_sales["Sales"].sum(), 2))
```

```python
df_sales["InvoiceDate"] = pd.to_datetime(
    df_sales["InvoiceDate"]
)

df_sales["Month"] = (
    df_sales["InvoiceDate"].dt.to_period("M")
)

monthly_sales = (
    df_sales.groupby("Month")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Orders=("Invoice", "nunique"),
        Total_Quantity=("Quantity", "sum")
    )
    .reset_index()
)

monthly_sales["Month"] = (
    monthly_sales["Month"].astype(str)
)

display(monthly_sales)

print("Highest Sales Month:")
display(
    monthly_sales.loc[
        monthly_sales["Total_Sales"].idxmax()
    ]
)
```

```python
df_returns = df_clean[
    (df_clean["Quantity"] < 0)
].copy()

df_returns["Return_Value"] = (
    df_returns["Quantity"].abs() *
    df_returns["Price"]
)

print("Total Return Transactions:", len(df_returns))

print("Total Returned Quantity:",
      df_returns["Quantity"].abs().sum())

print("Estimated Return Value:",
      round(df_returns["Return_Value"].sum(), 2))

print("\nCancellation Invoice Count:")
print(
    df_returns["Invoice"]
    .astype(str)
    .str.startswith("C")
    .sum()
)
```

```python
from sklearn.ensemble import IsolationForest

# Select transaction features
features = df_sales[
    ["Quantity", "Price", "Sales"]
].copy()

# Train anomaly detection model
model = IsolationForest(
    contamination=0.02,
    random_state=42
)

df_sales["Anomaly"] = model.fit_predict(features)

# -1 = unusual transaction
#  1 = normal transaction

anomalies = df_sales[
    df_sales["Anomaly"] == -1
].copy()

print("Total transactions:", len(df_sales))
print("Potential anomalies:", len(anomalies))

print("\nSample unusual transactions:")
display(
    anomalies[
        ["Invoice", "Description", "Quantity", "Price", "Sales"]
    ].head(10)
)
```

```python
anomaly_summary = (
    anomalies[
        ["Invoice", "Description", "Quantity", "Price", "Sales"]
    ]
    .sort_values("Sales", ascending=False)
)

print("Top 10 unusual transactions by sales value:")
display(anomaly_summary.head(10))

print("\nAnomaly Sales Summary:")
print("Total anomaly sales value:",
      round(anomalies["Sales"].sum(), 2))

print("Average anomaly transaction value:",
      round(anomalies["Sales"].mean(), 2))

print("Maximum anomaly transaction value:",
      round(anomalies["Sales"].max(), 2))
```

```python
import pandas as pd

df_rfm = df_sales.dropna(subset=["Customer ID"]).copy()

df_rfm["InvoiceDate"] = pd.to_datetime(
    df_rfm["InvoiceDate"]
)

reference_date = (
    df_rfm["InvoiceDate"].max() + pd.Timedelta(days=1)
)

rfm = df_rfm.groupby("Customer ID").agg(
    Recency=("InvoiceDate", lambda x:
             (reference_date - x.max()).days),
    Frequency=("Invoice", "nunique"),
    Monetary=("Sales", "sum")
).reset_index()

print("Customer count:", len(rfm))
display(rfm.head(10))
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

import numpy as np

# Select RFM features
rfm_features = rfm[["Recency", "Frequency", "Monetary"]].copy()

# Reduce skewness in Frequency and Monetary
rfm_features[["Frequency", "Monetary"]] = np.log1p(
    rfm_features[["Frequency", "Monetary"]].clip(lower=0)
)

# Scale the features
scaler = StandardScaler()
rfm_scaled = scaler.fit_transform(rfm_features)

# Train K-Means model
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

rfm["Customer_Segment"] = kmeans.fit_predict(rfm_scaled)

print("Customer segmentation completed!")
print("\nCustomers in each segment:")
print(rfm["Customer_Segment"].value_counts().sort_index())

display(rfm.head(10))
```

```python
segment_summary = (
    rfm.groupby("Customer_Segment")
    .agg(
        Customer_Count=("Customer ID", "count"),
        Avg_Recency=("Recency", "mean"),
        Avg_Frequency=("Frequency", "mean"),
        Avg_Monetary=("Monetary", "mean")
    )
    .round(2)
    .reset_index()
)

print("Customer Segment Summary:")
display(segment_summary)
```

```python
segment_comparison = (
    rfm.groupby("Customer_Segment")
    .agg(
        Customers=("Customer ID", "count"),
        Avg_Recency=("Recency", "mean"),
        Avg_Frequency=("Frequency", "mean"),
        Avg_Monetary=("Monetary", "mean")
    )
    .round(2)
)

display(segment_comparison)

# Save the segment summary
segment_comparison.to_csv(
    "customer_segment_summary.csv"
)

print("Customer segment summary saved successfully!")
```

```python
product_performance = (
    df_sales.groupby("Description")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Quantity=("Quantity", "sum"),
        Order_Count=("Invoice", "nunique")
    )
    .reset_index()
)

# Top 10 products by revenue
top_revenue_products = product_performance.nlargest(
    10, "Total_Sales"
)

print("Top 10 Products by Revenue:")
display(top_revenue_products)

# Top 10 products by quantity sold
top_quantity_products = product_performance.nlargest(
    10, "Total_Quantity"
)

print("\nTop 10 Products by Quantity Sold:")
display(top_quantity_products)
```

```python
product_contribution = (
    df_sales.groupby("Description")["Sales"]
    .sum()
    .reset_index()
)

total_sales = product_contribution["Sales"].sum()

product_contribution["Sales_Percentage"] = (
    product_contribution["Sales"] / total_sales * 100
).round(2)

product_contribution = product_contribution.sort_values(
    "Sales_Percentage", ascending=False
)

print("Total Sales:", round(total_sales, 2))

print("\nTop 10 Products by Sales Contribution:")
display(product_contribution.head(10))

print(
    "\nTop 10 contribution (%):",
    product_contribution.head(10)["Sales_Percentage"].sum()
)
```

```python
df_sales["InvoiceDate"] = pd.to_datetime(
    df_sales["InvoiceDate"]
)

df_sales["Month"] = (
    df_sales["InvoiceDate"].dt.to_period("M")
)

monthly_growth = (
    df_sales.groupby("Month")
    .agg(Total_Sales=("Sales", "sum"))
    .reset_index()
    .sort_values("Month")
)

monthly_growth["Previous_Month_Sales"] = (
    monthly_growth["Total_Sales"].shift(1)
)

monthly_growth["Growth_Percentage"] = (
    monthly_growth["Total_Sales"]
    .pct_change()
    .mul(100)
    .round(2)
)

monthly_growth["Month"] = (
    monthly_growth["Month"].astype(str)
)

display(monthly_growth)
```

```python
customer_data = df_sales.dropna(
    subset=["Customer ID"]
).copy()

# Total spending by each customer
customer_summary = (
    customer_data.groupby("Customer ID")
    .agg(
        Total_Spending=("Sales", "sum"),
        Total_Orders=("Invoice", "nunique"),
        Total_Quantity=("Quantity", "sum")
    )
    .reset_index()
)

# Average order value per customer
customer_summary["Average_Order_Value"] = (
    customer_summary["Total_Spending"] /
    customer_summary["Total_Orders"]
).round(2)

print("Customer Summary:")
display(customer_summary.head(10))

print("\nOverall Customer Metrics:")
print("Unique customers:", customer_summary["Customer ID"].nunique())
print("Average customer spending:",
      round(customer_summary["Total_Spending"].mean(), 2))
print("Average orders per customer:",
      round(customer_summary["Total_Orders"].mean(), 2))
```

```python
customer_type = customer_summary.copy()

customer_type["Customer_Type"] = customer_type[
    "Total_Orders"
].apply(
    lambda x: "One-Time Customer" if x == 1
    else "Repeat Customer"
)

customer_type_summary = (
    customer_type.groupby("Customer_Type")
    .agg(
        Customer_Count=("Customer ID", "count"),
        Total_Sales=("Total_Spending", "sum")
    )
    .reset_index()
)

customer_type_summary["Customer_Percentage"] = (
    customer_type_summary["Customer_Count"]
    / customer_type_summary["Customer_Count"].sum()
    * 100
).round(2)

print("Customer Type Analysis:")
display(customer_type_summary)
```

```python
#Customer Segmentation using KMeans

import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

df_rfm = df_sales.dropna(
    subset=["Customer ID"]
).copy()

df_rfm["InvoiceDate"] = pd.to_datetime(
    df_rfm["InvoiceDate"]
)

reference_date = (
    df_rfm["InvoiceDate"].max()
    + pd.Timedelta(days=1)
)

rfm = df_rfm.groupby("Customer ID").agg(
    Recency=("InvoiceDate",
             lambda x: (reference_date - x.max()).days),
    Frequency=("Invoice", "nunique"),
    Monetary=("Sales", "sum")
).reset_index()

rfm_features = rfm[
    ["Recency", "Frequency", "Monetary"]
].copy()

rfm_features[["Frequency", "Monetary"]] = np.log1p(
    rfm_features[["Frequency", "Monetary"]].clip(lower=0)
)

scaler = StandardScaler()
rfm_scaled = scaler.fit_transform(rfm_features)

kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

rfm["Customer_Segment"] = kmeans.fit_predict(
    rfm_scaled
)

display(
    rfm.groupby("Customer_Segment").agg(
        Customer_Count=("Customer ID", "count"),
        Avg_Recency=("Recency", "mean"),
        Avg_Frequency=("Frequency", "mean"),
        Avg_Monetary=("Monetary", "mean")
    ).round(2)
)
```

```python
monthly_query = """
SELECT
    YEAR(InvoiceDate) AS Sales_Year,
    MONTH(InvoiceDate) AS Sales_Month_Number,
    ROUND(SUM(Quantity * Price), 2) AS Total_Sales,
    COUNT(DISTINCT Invoice) AS Total_Orders
FROM retail_analysis
WHERE Quantity > 0
  AND Price > 0
GROUP BY YEAR(InvoiceDate), MONTH(InvoiceDate)
ORDER BY Sales_Year, Sales_Month_Number;
"""

monthly_sales_sql = pd.read_sql(monthly_query, engine)
display(monthly_sales_sql)
```

```python
query = """
SELECT
    CASE
        WHEN Order_Count = 1 THEN 'One-Time Customer'
        ELSE 'Repeat Customer'
    END AS Customer_Type,
    COUNT(*) AS Total_Customers
FROM (
    SELECT
        `Customer ID`,
        COUNT(DISTINCT Invoice) AS Order_Count
    FROM retail_analysis
    WHERE `Customer ID` IS NOT NULL
      AND Quantity > 0
      AND Price > 0
    GROUP BY `Customer ID`
) AS customer_orders
GROUP BY Customer_Type;
"""

customer_type_sql = pd.read_sql(query, engine)
display(customer_type_sql)
```

```python
query = """
SELECT
    `Customer ID`,
    COUNT(DISTINCT Invoice) AS Total_Orders,
    ROUND(SUM(Quantity * Price), 2) AS Total_Sales
FROM retail_analysis
WHERE `Customer ID` IS NOT NULL
  AND Quantity > 0
  AND Price > 0
GROUP BY `Customer ID`
ORDER BY Total_Sales DESC
LIMIT 10;
"""

top_customers_sql = pd.read_sql(query, engine)
display(top_customers_sql)
```

```python
query = """
SELECT
    Description,
    SUM(Quantity) AS Total_Quantity_Sold,
    ROUND(SUM(Quantity * Price), 2) AS Total_Revenue,
    ROUND(AVG(Price), 2) AS Average_Price
FROM retail_analysis
WHERE Quantity > 0
  AND Price > 0
  AND Description IS NOT NULL
GROUP BY Description
ORDER BY Total_Revenue DESC
LIMIT 15;
"""

product_performance_sql = pd.read_sql(query, engine)
display(product_performance_sql)
```

```python
customer_revenue_query = """
SELECT
    `Customer ID`,
    ROUND(SUM(Quantity * Price), 2) AS Customer_Revenue,
    ROUND(
        SUM(Quantity * Price) * 100 /
        (SELECT SUM(Quantity * Price)
         FROM retail_analysis
         WHERE Quantity > 0 AND Price > 0),
        2
    ) AS Revenue_Contribution_Percentage
FROM retail_analysis
WHERE `Customer ID` IS NOT NULL
  AND Quantity > 0
  AND Price > 0
GROUP BY `Customer ID`
ORDER BY Customer_Revenue DESC
LIMIT 10;
"""

customer_revenue_sql = pd.read_sql(customer_revenue_query, engine)
display(customer_revenue_sql)
```

```python
order_status_query = """
SELECT
    CASE
        WHEN LEFT(Invoice, 1) = 'C'
        THEN 'Cancelled / Returned'
        ELSE 'Regular Invoice'
    END AS Order_Status,
    COUNT(DISTINCT Invoice) AS Total_Invoices,
    ROUND(SUM(Quantity * Price), 2) AS Total_Amount
FROM retail_analysis
GROUP BY Order_Status;
"""

order_status_sql = pd.read_sql(order_status_query, engine)
display(order_status_sql)
```

```python
aov_query = """
SELECT
    COUNT(*) AS Total_Orders,
    ROUND(SUM(Order_Value), 2) AS Total_Revenue,
    ROUND(AVG(Order_Value), 2) AS Average_Order_Value
FROM (
    SELECT
        Invoice,
        SUM(Quantity * Price) AS Order_Value
    FROM retail_analysis
    WHERE Quantity > 0
      AND Price > 0
      AND LEFT(Invoice, 1) <> 'C'
    GROUP BY Invoice
) AS order_summary;
"""

aov_sql = pd.read_sql(aov_query, engine)
display(aov_sql)
```

```python
new_customer_query = """
SELECT
    YEAR(First_Order_Date) AS Purchase_Year,
    MONTH(First_Order_Date) AS Purchase_Month,
    COUNT(*) AS New_Customers
FROM (
    SELECT
        `Customer ID`,
        MIN(InvoiceDate) AS First_Order_Date
    FROM retail_analysis
    WHERE `Customer ID` IS NOT NULL
      AND Quantity > 0
      AND Price > 0
    GROUP BY `Customer ID`
) AS customer_first_purchase
GROUP BY
    YEAR(First_Order_Date),
    MONTH(First_Order_Date)
ORDER BY
    YEAR(First_Order_Date),
    MONTH(First_Order_Date);
"""

new_customer_sql = pd.read_sql(new_customer_query, engine)
display(new_customer_sql)
```

```python
country_contribution_query = """
SELECT
    Country,
    ROUND(SUM(Quantity * Price), 2) AS Country_Sales,
    ROUND(
        SUM(Quantity * Price) * 100 /
        (SELECT SUM(Quantity * Price)
         FROM retail_analysis
         WHERE Quantity > 0 AND Price > 0),
        2
    ) AS Sales_Contribution_Percentage
FROM retail_analysis
WHERE Quantity > 0
  AND Price > 0
GROUP BY Country
ORDER BY Country_Sales DESC
LIMIT 15;
"""

country_contribution_sql = pd.read_sql(
    country_contribution_query, engine
)
display(country_contribution_sql)
```

```python
order_value_query = """
SELECT
    CASE
        WHEN Order_Value >= 500 THEN 'High Value'
        WHEN Order_Value >= 100 THEN 'Medium Value'
        ELSE 'Low Value'
    END AS Order_Category,
    COUNT(*) AS Total_Orders,
    ROUND(SUM(Order_Value), 2) AS Total_Revenue,
    ROUND(AVG(Order_Value), 2) AS Average_Order_Value
FROM (
    SELECT
        Invoice,
        SUM(Quantity * Price) AS Order_Value
    FROM retail_analysis
    WHERE Quantity > 0
      AND Price > 0
      AND LEFT(Invoice, 1) <> 'C'
    GROUP BY Invoice
) AS order_summary
GROUP BY Order_Category
ORDER BY Total_Revenue DESC;
"""

order_value_sql = pd.read_sql(order_value_query, engine)
display(order_value_sql)
```

```python
quality_query = """
SELECT
    COUNT(*) AS Total_Rows,
    SUM(CASE WHEN `Customer ID` IS NULL THEN 1 ELSE 0 END)
        AS Missing_Customer_ID,
    SUM(CASE WHEN Description IS NULL THEN 1 ELSE 0 END)
        AS Missing_Description,
    SUM(CASE WHEN InvoiceDate IS NULL THEN 1 ELSE 0 END)
        AS Missing_InvoiceDate,
    SUM(CASE WHEN Quantity <= 0 THEN 1 ELSE 0 END)
        AS Non_Positive_Quantity,
    SUM(CASE WHEN Price <= 0 THEN 1 ELSE 0 END)
        AS Non_Positive_Price
FROM retail_analysis;
"""

quality_sql = pd.read_sql(quality_query, engine)
display(quality_sql)
```

```python
quality_query = """
SELECT
    COUNT(*) AS Total_Rows,
    SUM(CASE WHEN `Customer ID` IS NULL THEN 1 ELSE 0 END)
        AS Missing_Customer_ID,
    SUM(CASE WHEN Description IS NULL THEN 1 ELSE 0 END)
        AS Missing_Description,
    SUM(CASE WHEN InvoiceDate IS NULL THEN 1 ELSE 0 END)
        AS Missing_InvoiceDate,
    SUM(CASE WHEN Quantity <= 0 THEN 1 ELSE 0 END)
        AS Non_Positive_Quantity,
    SUM(CASE WHEN Price <= 0 THEN 1 ELSE 0 END)
        AS Non_Positive_Price
FROM retail_analysis;
"""

quality_sql = pd.read_sql(quality_query, engine)
display(quality_sql)
```

```python
summary_query = """
SELECT
    ROUND(SUM(Quantity * Price), 2) AS Total_Revenue,
    COUNT(DISTINCT Invoice) AS Total_Invoices,
    COUNT(DISTINCT `Customer ID`) AS Total_Customers,
    COUNT(DISTINCT StockCode) AS Unique_Products,
    ROUND(AVG(Price), 2) AS Average_Product_Price,
    SUM(Quantity) AS Total_Quantity_Sold
FROM retail_analysis
WHERE Quantity > 0
  AND Price > 0;
"""

sql_summary = pd.read_sql(summary_query, engine)
display(sql_summary)
```
